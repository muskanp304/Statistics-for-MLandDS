```python
#load the datset
df = pd.read_csv("C:/Users/temp.admin/Documents/Stats AIDS/diabetes.csv")
df.head
```

```python
#Select the variable and divide into groups
non_diabetic= df[df["Outcome"] ==0]["Glucose"]
diabetic =df[df["Outcome"]==1]["Glucose"]
print("Non-diabetic Patient :",len(non_diabetic))
print("Diabetic Patient :",len(diabetic))
```

```python
#Calculate Point Estimates
#sample mean
sample_mean =df["Glucose"].mean()
print("Sample Mean of Glucose :", sample_mean)
```

```python
#Sample proportion
diabetic_count =(df["Outcome"]==1).sum()
total_count = len(df)
sample_proportion = diabetic_count/total_count
print("Sample proportion :",sample_proportion)
print("sample Proportion(%):",sample_proportion*100)
```

```python
# Construct Confidence intervals
from scipy.stats import t

mean = df["Glucose"].mean()
std = df["Glucose"].std()
n = len(df)

confidence_level = 0.95
alpha = 1 - confidence_level

t_critical = t.ppf(1 - alpha/2, n - 1)

margin_error = t_critical * (std / np.sqrt(n))

lower = mean - margin_error
upper = mean + margin_error

print("95% Confidence Interval for Mean Glucose")
print("Lower Limit:", lower)
print("Upper Limit:", upper)
```

Formulate Null and Alternative Hypothesis

```python
# Research Question 1

# Is the average glucose level of patients significantly different from 120?

# Null Hypothesis — H₀
# The population mean glucose level is equal to 120.
# H₀: μ = 120

# Alternative Hypothesis — H₁
# The population mean glucose level is different from 120.
# H₁: μ ≠ 120
# We use 120 only as a reference value for the one-sample test.
```

```python
# One-Sample t-Test
from scipy.stats import ttest_1samp

t_statistic, p_value = ttest_1samp(df["Glucose"], 120)

print("One-Sample t-Test")
print("Test Statistic:", t_statistic)
print("P-value:", p_value)
```

```python
#Compare p-value with Significance level
alpha = 0.05
print("Significance Level:", alpha)
print("P-value:", p_value)

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

```python
#Two sample T-test
from scipy.stats import ttest_ind

t_statistic, p_value = ttest_ind(
    diabetic,
    non_diabetic,
    equal_var=False
)

print("Two-Sample t-Test")
print("Test Statistic:", t_statistic)
print("P-value:", p_value)
```

```python
#Compare 
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

```python
#Create BMI Categories
df["BMI_Category"] = np.where(
    df["BMI"] < 25,
    "Normal",
    "High"
)

df[["BMI", "BMI_Category", "Outcome"]].head()
```

```python
#Create table
table = pd.crosstab(
    df["BMI_Category"],
    df["Outcome"]
)

print(table)
```

```python
#Chi-Square Test
from scipy.stats import chi2_contingency

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-Square Test")
print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
```

```python
#Compare 
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```
