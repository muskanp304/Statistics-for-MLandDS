Q4. Distance Measures and Scaling

Create a small dataset containing at least four observations and two numerical features
with diƯerent scales. Apply standardization and calculate the Euclidean distance
between two observations before and after standardization. 


```python
from  sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import euclidean
import pandas as pd 
import numpy as np
```

```python
df = pd.DataFrame({
    "Study Hours " : [2,4,6,8,10],
    "Marks" : [40,60,80,100,120]
})
df.head()
```

```python
d1 = euclidean(df.iloc[0], df.iloc[1])
print("Distance before euclidean", d1)
```

```python
scalar = StandardScaler()
scalar_data = scalar.fit_transform(df)
scalar_df = pd.DataFrame(scalar_data ,columns=df.columns)
scalar_df.head
```

```python
d2 = euclidean(scalar_df.iloc[0], scalar_df.iloc[1])
print("Distance after euclidean", d2)
```
