Q6. Bootstrap Confidence Interval

Using the dataset 12, 15, 14, 10, 18, 16, 13, 17, 11, 14, implement bootstrap resampling
to estimate a 95% confidence interval for the sample mean. 

```python
import numpy as np
```

```python
data =[12,15,14,10,18,16,13,17,11,14]
```

```python
print("sample Mean ",np.mean(data))
```

```python
np.random.seed(42)
bootstrp_means=[]

for i in range(1000):
    sample = np.random.choice(data, size=len(data), replace = True)
    bootstrp_means.append(np.mean(sample))
```

```python
lower = np.percentile(bootstrp_means, 2.5)
upper = np.percentile(bootstrp_means, 97.5)
```

```python
print("95% Confidence Interval :",lower ,"to ", upper)
```
