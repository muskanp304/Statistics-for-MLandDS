Q12. PCA Visualization

Apply PCA to the standardized numerical features of the Iris dataset and reduce them to
two principal components. Generate a two-dimensional scatter plot of the transformed
observations, using the actual species labels to distinguish the points. 


```python
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import  matplotlib.pyplot as plt 
import pandas as pd 
```

```python
iris=load_iris()
```

```python
x=iris.data
y=iris.target
```

```python
scalar = StandardScaler()
scaled = scalar.fit_transform(x)
```

```python
pca = PCA(n_components=2)
x_pca = pca.fit_transform(scaled)
```

```python
plt.scatter(x_pca[:,0], x_pca[:,1], c=y)
plt.xlabel("Principal component 1")
plt.ylabel("Principle component 2")
plt.title("PCA visulization of IRIS dataset")
plt.show()
```
