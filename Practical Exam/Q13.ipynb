Q13. Cross-Validation
Using the Breast Cancer Wisconsin dataset available through scikit-learn, train a Logistic
Regression classifier and evaluate it using 5-fold cross-validation. Report the individual
fold scores and mean accuracy. 


```python
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import cross_val_score
import pandas as pd 
```

```python
cancer = load_breast_cancer()
```

```python
x=cancer.data
y=cancer.target
```

```python
model =make_pipeline(
    StandardScaler(),
    LogisticRegression()
)
```

```python
scores = cross_val_score(
    model, x, y, cv=5, scoring="accuracy"
)
```

```python
print("Mean Accuracy:", scores.mean())
print("Mean Accuracy (%):", scores.mean() * 100)
```
