Q1. Iris Dataset — Exploratory Data Analysis

Using the Iris dataset, write a Python program to calculate the mean, median and
standard deviation of the four numerical features and generate a box plot for one selected
feature. 


```python
from sklearn.datasets import load_iris
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
iris = load_iris()
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
print(df.mean())
```

```python
df.median()
```

```python
df.std()
```

```python
plt.boxplot(df["sepal length (cm)"], patch_artist=True)
plt.title("Box plot of Sepal length")
plt.show()
```

```python
from sklearn.datasets import load_iris
import pandas as pd 
import matplotlib.pyplot as plt 

iris = load_iris()
df = pd.DataFrame(iris.data , columns=iris.feature_names)
print("Mean\n")
print(df.mean())
print("Median\n")
print(df.median())
print("Standard Deviation\n")
print(df.std())

plt.boxplot(df["sepal length (cm)"], patch_artist=True)
plt.title("Box plot of Sepal length")
plt.show()
```
