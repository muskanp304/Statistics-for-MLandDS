Q2. Diabetes Dataset — Distribution Analysis

Using load_diabetes() from scikit-learn, calculate the descriptive statistics of the BMI
feature and generate a histogram to examine its distribution. 

```python
from sklearn.datasets import load_diabetes
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
diabetes = load_diabetes()
df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
```

```python
df["bmi"].describe()
```

```python
plt.hist(df["bmi"],edgecolor="black")
plt.title("BMI distribution")
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.show()
```
