Q5. One-Sample t-Test

Consider the observations 52, 48, 51, 49, 54, 50, 53, 47. Perform a one-sample t-test to
determine whether the population mean diƯers significantly from 50 at the 5%
significance level. 


```python
from scipy.stats import ttest_1samp
```

```python
data =[52,48,51,49,54,50,53,47]
```

```python
t_stastitics, p_value = ttest_1samp(data,50)
print("t-Statistics Value",t_stastitics)
print("p-value",p_value)
```

```python
if p_value < 0.05 :
    print(" Reject the null hypothesis")
else :
    print("Fail to reject the null hypothesis")
```
