Q3. Iris Dataset — Correlation Analysis

Calculate the Pearson correlation matrix of the numerical features in the Iris dataset and
visualize it using a heatmap. Identify the feature pair with the strongest absolute
correlation. 

```python
from sklearn.datasets import load_iris
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
iris = load_iris()
df = pd.DataFrame(iris.data , columns=iris.feature_names)
df.head
```

```python
corr = df.corr(method="pearson")
print(corr)
```

```python
import seaborn as sns
```

```python
sns.heatmap(corr, annot=True)
plt.title(" correlation Heatmap")
plt.show()
```

```python
# wrong beacuse giving the diagoanl value

# Find the strongest absolute correlation
corr_abs = corr.abs()
corr_abs.values[[range(len(corr_abs))]*2] = 0
pair = corr_abs.unstack().idxmax()

print("Strongest correlated pair:", pair)
print("Correlation value:", corr.loc[pair[0], pair[1]])
```

```python
# Find the strongest absolute correlation
corr_abs = corr.abs()

# Ignore diagonal values
import numpy as np
np.fill_diagonal(corr_abs.values, 0)

# Find the strongest correlation
pair = corr_abs.unstack().idxmax()

print("Strongest correlated pair:", pair)
print("Correlation value:", corr.loc[pair[0], pair[1]])
```
