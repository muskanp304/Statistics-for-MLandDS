Q8. Regression Model Comparison

Using the Diabetes dataset, train Linear Regression and Ridge Regression models using
the same train-test split. Compare their test-set R2
 scores.

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
import pandas as pd 
```

```python
diabetes = load_diabetes()
df = pd.DataFrame(diabetes.data , columns=diabetes.feature_names)
df.head
```

```python
df["target"] = diabetes.target
```

```python
x=df[["bmi"]]
y=df["target"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
linear_regression = LinearRegression()
linear_regression.fit(x_train, y_train)
```

```python
ridge_regression = Ridge()
ridge_regression.fit(x_train, y_train)
```

```python
pred_linear = linear_regression.predict(x_test)
pred_ridge = ridge_regression.predict(x_test)
```

```python
print("r2-Score of Linear Reggression ", r2_score(y_test, pred_linear))
print("r2-Score of Ridge Reggression ", r2_score(y_test, pred_ridge))
```
