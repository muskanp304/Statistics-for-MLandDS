Q4. Distance Measures and Scaling

Create a small dataset containing at least four observations and two numerical features
with diƯerent scales. Apply standardization and calculate the Euclidean distance
between two observations before and after standardization. 


```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import euclidean
```

```python
df = pd.DataFrame({"study_Hours":[2,4,6,8], "Marks":[40,80,60,90]})
print(df)
```

```python
d1 = euclidean(df.iloc[0], df.iloc[1])
print("Euclidean Distance before standardization",d1)
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
```

```python
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)
```

```python
print(scaled_df)
```

```python
d2 = euclidean(scaled_df.iloc[0], scaled_df.iloc[1])
print("Euclidean distance after standardization ", d2)
```

```python
import pandas as pd 
from  sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import euclidean

df = pd.DataFrame({
    "Study Hours" : [2,4,6,8],
    "Marks" : [40,60,80,100]
})

print("original data",df)

d1 = euclidean(df.iloc[0],df.iloc[1])
print("Euclidean distance before standardization",d1)

scaler = StandardScaler()
scaler_data = scaler.fit_transform(df)
scaler_df = pd.DataFrame(scaler_data, columns=df.columns)
print("Scaled data",scaler_df)

d2 = euclidean(scaler_df.iloc[0], scaler_df.iloc[1])
print("euclidean distance After Standardization",d2)
```
